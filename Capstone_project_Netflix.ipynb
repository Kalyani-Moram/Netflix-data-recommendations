```python
# Manual Filteration
# Nneed to get the count of every customer with respect to rating.
data_cust_summary = df.groupby('CustID')['Ratings'].count()
data_cust_summary
```

```python
cust_benchmark = round(data_cust_summary.quantile(0.60))
# I will select only those customers who have seen more than 36 movies.
cust_benchmark
```

```python
Rejected_cust = data_cust_summary[data_cust_summary<cust_benchmark].index
# When we use groupby fxn the column becomes index
Rejected_cust
```

```python
# Filtering our movie data
# First we will get the count of ratings for every movie.
data_movie_summary = df.groupby('MovieID')['Ratings'].count()
data_movie_summary
```

```python
movie_benchmark = round(data_movie_summary.quantile(0.60))
movie_benchmark
```

```python
Rejected_movies = data_movie_summary[data_movie_summary<movie_benchmark].index
Rejected_movies
```

```python
# Now we have selected the customers and movies that we need to reject.
# Now we will update the result in our actual dataframe (removing these rejected values)
```

```python
df[df['MovieID'].isin(Rejected_movies)]  # Here are all the movies that we need to reject
```

```python
df = df[~df['MovieID'].isin(Rejected_movies)]
df
```

```python
df = df[~df['CustID'].isin(Rejected_cust)]
df
```

**Recommendation**

Recommendation for any particular user for all those movie which are not rated by him.

```python
df.head()
```

```python
df_title.head()
```

```python
# Creating a copy of the df_title DataFrame for user 712664
user_712664 = df_title.copy()

# Check the copied DataFrame
print(user_712664.head())
```

```python
# identifying the unrated movies of user id 712664

# Step 1: Define the user ID and identify rated movies
user_id = 712664

# Movies the user has already rated
rated_movies = df[df['CustID'] == user_id]['MovieID'].unique()

# Step 2: Find unrated movies (those that user hasn't rated yet)
unrated_movies = df_title[~df_title['MovieID'].isin(rated_movies)]

# Step 3: Predict ratings for each unrated movie
predictions = []
for movie_id in unrated_movies['MovieID']:
    pred = model.predict(user_id, movie_id)
    predictions.append((movie_id, pred.est))

# Step 4: Create a DataFrame from the predictions
pred_df = pd.DataFrame(predictions, columns=['MovieID', 'PredictedRating'])

# Step 5: Merge predictions with movie details (movie name and year)
user_712664 = df_title.merge(pred_df, on='MovieID')

# Step 6: Sort by predicted ratings and recommend top 10 movies
top_recommendations = user_712664.sort_values('PredictedRating', ascending=False).head(10)

# Step 7: Display the top recommended movies
print(top_recommendations[['Name', 'Year', 'PredictedRating']])
```

```python
# recommendation for any particular user for all those movie.

user_1488844 = df_title.copy()
```

```python
user_1488844['Estimate Score'] = user_1488844['MovieID'].apply(lambda x : model.predict(1488844,x).est)
```
